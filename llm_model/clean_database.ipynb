```python
df = df.drop_duplicates()
```

```python
df = df[df['abstract'].str.len() != 1]
```

```python
df.info()
```

```python
df = df.dropna(subset=['abstract'])
```

```python
df.info()
```

```python
with open ('chosen.txt', 'r') as file:
    chosen = file.readlines()
```

```python
chosen_clean = list()
for line in chosen:
    chosen_clean.append(line.strip())
```

```python
chosen_clean = [line for line in chosen_clean if line != '']
```

```python
chosen_clean
```

```python
len(chosen_clean)
```

```python
def check_chosen(row, chosen_list):
    for title in chosen_list:
        if title in row['title']:
            return True
    return False
```

```python
df['chosen'] = df.apply(lambda row: check_chosen(row, chosen_clean), axis=1)
```
